# **Queue**

```python
class Queue:
    def __init__(self,capacity):
        self.queue = []
        self.capacity = capacity
        
    def is_empty(self):
        return len(self.queue) == 0
    
    def is_full(self):
        return len(self.queue) == self.capacity
    
    def en_queue(self,item):
        if self.is_full():
            print(" Queue is FULL ")
        else:
            self.queue.append(item)
        
    def de_queue(self):
        if self.is_empty():
            print(" Queue is EMPTY ")
            return None
        else:
            return self.queue.pop(0)
        
    def peek(self):
        if self.is_empty():
            print(" Queue is EMPTY ")
            return None
        else:
            return self.queue[0]
        
    def print(self):
        return [" ".join(map(str,self.queue))]
```

```python
q = Queue(5)
q.en_queue(2)
q.en_queue(4)
q.en_queue(5)
q.en_queue(5)
q.en_queue(1)

q.de_queue()
q.de_queue()
q.de_queue()
q.de_queue()
q.de_queue()
q.print()
```

# **Circular Queue**

```python
class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.queue = [None] * capacity
        self.front = 0
        self.rear = 0
        self.size = 0

    # --- CORE OPERATIONS ---

    def enqueue(self, item):
        if self.is_full():
            print("Queue Overflow!")
            return "Full"
        
        self.queue[self.rear] = item
        # Move rear forward and wrap around if necessary
        self.rear = (self.rear + 1) % self.capacity
        self.size += 1
        print(f"Enqueued: {item}")

    def dequeue(self):
        if self.is_empty():
            print("Queue Underflow!")
            return "Empty"
        
        data = self.queue[self.front]
        self.queue[self.front] = None # Optional: Clear for visualization
        # Move front forward and wrap around if necessary
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return data

    # --- UTILITY FUNCTIONS ---

    def peek(self):
        """Look at the element at the front without removing it."""
        if self.is_empty():
            return "Queue is Empty"
        return self.queue[self.front]

    def is_full(self):
        """Check if queue has reached capacity."""
        return self.size == self.capacity

    def is_empty(self):
        """Check if queue has zero items."""
        return self.size == 0

    def get_size(self):
        """Return current number of items."""
        return self.size

    def show(self):
        """Print the raw array and pointer positions."""
        print(f"\nData: {self.queue}")
        print(f"Front Index: {self.front} | Rear Index: {self.rear} | Current Size: {self.size}")
        
    def Front(self) -> int:
        return self.queue[self.front] if not self.is_empty() else -1

    def Rear(self) -> int:
        if self.is_empty():
            return -1
        
        # FIXED: Look at the index one step behind 'rear'
        # We add capacity before modulo to handle the case where rear is 0
        last_index = (self.rear - 1 + self.capacity) % self.capacity
        return self.queue[last_index]
```

```python
cq = CircularQueue(5)
cq.enqueue(1)
cq.enqueue(2)
cq.enqueue(3)
cq.enqueue(4)
cq.enqueue(5)
cq.dequeue()
cq.dequeue()
cq.dequeue()
cq.dequeue()
cq.dequeue()
cq.enqueue(5)
cq.show()
cq.Front()
cq.Rear()
```

# **Deque**

### A Simple Deque (Double-Ended Queue) is a more flexible version of a standard queue. While a standard queue follows the "First-In, First-Out" (FIFO) rule, a Deque allows you to add and remove items from both ends.

```python
class SimpleDeque:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    # --- REAR Operations (Back) ---
    def add_rear(self, item):
        self.items.append(item)

    def remove_rear(self):
        if not self.is_empty():
            return self.items.pop()
        return "Deque is empty"

    # --- FRONT Operations (Beginning) ---
    def add_front(self, item):
        self.items.insert(0, item)

    def remove_front(self):
        if not self.is_empty():
            return self.items.pop(0)
        return "Deque is empty"

    def size(self):
        return len(self.items)

    def show(self):
        print("Front ->", self.items, "<- Rear")
```

# Circular Deque

```python
class Deque:
    def __init__(self, capacity):
        self.capacity = capacity
        self.queue = [None] * capacity
        self.front = 0
        self.rear = 0
        self.size = 0

    def is_full(self):
        return self.size == self.capacity

    def is_empty(self):
        return self.size == 0

    # --- INSERT OPERATIONS ---

    def insert_rear(self, item):
        if self.is_full():
            print("Deque Full!")
            return
        self.queue[self.rear] = item
        self.rear = (self.rear + 1) % self.capacity
        self.size += 1

    def insert_front(self, item):
        if self.is_full():
            print("Deque Full!")
            return
        # Move front backward: if at 0, it wraps to capacity-1
        self.front = (self.front - 1 + self.capacity) % self.capacity
        self.queue[self.front] = item
        self.size += 1

    # --- DELETE OPERATIONS ---

    def delete_front(self):
        if self.is_empty():
            print("Deque Empty!")
            return None
        data = self.queue[self.front]
        self.queue[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return data

    def delete_rear(self):
        if self.is_empty():
            print("Deque Empty!")
            return None
        # Move rear backward to the actual last item
        self.rear = (self.rear - 1 + self.capacity) % self.capacity
        data = self.queue[self.rear]
        self.queue[self.rear] = None
        self.size -= 1
        return data

    def show(self):
        print(f"Data: {self.queue} | Front: {self.front} | Rear: {self.rear}")
```
