# **Binary Tree**

```python
from collections import deque
# defining node
class Node:
    def __init__(self,item):
        self.item=item
        self.left=None
        self.right=None
        
class BT:
    def __init__(self):
        self.root = None
        
    def insertValues(self,lst,i,n):
        def rinsertValues(lst,i,n):
            root = None
        
            if i < n:
                root = Node(lst[i])
                root.left = rinsertValues(lst,2*i+1,n)
                root.right = rinsertValues(lst,2*i+2,n)
            return root
        
        self.root = rinsertValues(lst,i,n)
    
    def inorder(self):
        result = []
        def rinorder(node):
            if node:
                rinorder(node.left)
                result.append(node.item)
                rinorder(node.right)
        
        rinorder(self.root)
        return result
    
    def level_order(self):
        def ilevel_order(root):
            result = []
            q = deque([root])

            while q:
                qlenth = len(q)
                level = []

                for i in range(qlenth):
                    node = q.popleft()
                    level.append(node.item)
                    if node.left:
                        q.append(node.left)
                    if node.right:
                        q.append(node.right)
                        
                result.append(level)
            return [item for sublist in result for item in sublist]
        return ilevel_order(self.root)
    
    # Checking if a tree is a mirror image of itself
    def symmetric(self):
        def isMirror(t1, t2):
            # Case 1: Both are empty (Symmetric)
            if not t1 and not t2:
                return True
            # Case 2: One is empty, or values don't match (Not Symmetric)
            if not t1 or not t2 or t1.item != t2.item:
                return False
            
            # Case 3: Compare "Outer" children and "Inner" children
            # Left of t1 with Right of t2 AND Right of t1 with Left of t2
            return isMirror(t1.left, t2.right) and isMirror(t1.right, t2.left)

        if not self.root:
            return True
        return isMirror(self.root.left, self.root.right)
    
    def max_node(self):
        def imax_node(root):
            queue = deque([root])
            max_node = float("-inf")
            while queue:
                node = queue.popleft()
                
                if node.left:
                    queue.append(node.left)
                    
                if node.right:
                    queue.append(node.right)
                    
                if max_node < node.item:
                    max_node = node.item
            return max_node            
        return imax_node(self.root)
```

```python
btree = BT()
a = [1,2,2,3,4,5,3]
n = len(a)
btree.insertValues(a,0,n)
btree.inorder()
# btree.symmetric()
btree.level_order()
btree.max_node()
```

# **Binary Search Tree**

```python
# Definition of the Node class
class Node:
    def __init__(self, item=None, left=None, right=None):
        self.item = item   # The data value stored in the node
        self.left = left   # Pointer to the left child (smaller values)
        self.right = right # Pointer to the right child (larger values)

# Definition of the Binary Search Tree class
class BST:
    def __init__(self):
        self.root = None # Initialize an empty tree
        
    # Public method to insert data (hides recursion details from user)
    def insert(self, data):
        def rinsert(root, data):
            # Base Case: Found an empty spot, create new node here
            if root is None:
                return Node(data)
            
            # Recursive Step: Decide to go Left or Right
            if data <= root.item:
                root.left = rinsert(root.left, data)
            # elif data > root.item:
            else:
                root.right = rinsert(root.right, data)
                
            # Return the (potentially updated) root pointer to the parent
            return root

        self.root = rinsert(self.root, data)
        
    # Public method to insert multiple data
    def insertValues(self, data):
        for i in data:
            self.insert(i)
        
    # Public method to search for data
    def search(self, data):
        def rsearch(root, data):
            # Base Case: Root is empty (Not Found) or Matches (Found)
            if root is None or root.item == data:
                return root
            
            # Recursive Step: Navigate down
            if data <= root.item:
                return rsearch(root.left, data)
            else:
                return rsearch(root.right, data)

        return rsearch(self.root, data)
    
    # Public method to delete data
    def delete(self, data):
        def min_value(temp):
            current = temp
            while current.left is not None:
                current = current.left
            return current.item

        def rdelete(root, data):
            # Base Case: Tree is empty or Item not found
            if root is None:
                return root
            
            # SEARCH PHASE: Navigate to find the node
            if data < root.item:
                root.left = rdelete(root.left, data)
            elif data > root.item:
                root.right = rdelete(root.right, data)
                
            # DELETION PHASE: Node found
            else:
                # Case 1: Node has no left child (or is a leaf)
                if root.left is None:
                    return root.right
                
                # Case 2: Node has no right child
                elif root.right is None:
                    return root.left
                
                # Case 3: Node has TWO children
                # Step A: Get the minimum value from the right subtree
                root.item = min_value(root.right)
                
                # Step B: Delete that duplicate minimum node from the right subtree
                root.right = rdelete(root.right, root.item)
                
            return root

        self.root = rdelete(self.root, data)
    
    # --- Traversal Methods ---
    
    # Inorder: Left -> Root -> Right (Results in sorted order)
    def inorder(self):
        lst = []
        def rinorder(root):
            if root:
                rinorder(root.left)   # Visit Left
                lst.append(root.item) # Visit Root
                rinorder(root.right)  # Visit Right
        
        rinorder(self.root)
        return lst
    
    # Preorder: Root -> Left -> Right (Useful for copying trees)
    def preorder(self):
        lst = []
        def rpreorder(root):
            if root:
                lst.append(root.item)   # Visit Root
                rpreorder(root.left)    # Visit Left
                rpreorder(root.right)   # Visit Right
        
        rpreorder(self.root)
        return lst
    
    # Postorder: Left -> Right -> Root (Useful for deleting trees)
    def postorder(self):
        lst = []
        def rpostorder(root):
            if root:
                rpostorder(root.left)   # Visit Left
                rpostorder(root.right)  # Visit Right
                lst.append(root.item)   # Visit Root
        
        rpostorder(self.root)
        return lst
                
    # Public method to get size
    def size(self):
        def rsize(root):
            if root is None:
                return 0
            else:
                # Count Self (1) + Count Left Subtree + Count Right Subtree
                return 1 + rsize(root.left) + rsize(root.right)
        
        return rsize(self.root)
        
    # height of a Nodes / maximum connected Nodes / depth of a Nodes
    def depthNodes(self):
        def rdepthNodes(root):
            if root is None:
                return 0
            return max(rdepthNodes(root.left), rdepthNodes(root.right)) + 1
        
        return rdepthNodes(self.root)
    
    # height of a Tree / maximum connected edges / depth of a tree
    def depthTree(self):
        def rdepthTree(root):
            if root is None:
                return -1
            
            left_h = rdepthTree(root.left)
            right_h = rdepthTree(root.right)

            return max(left_h, right_h) + 1
        
        return rdepthTree(self.root)
    
    # Inverting a subtree
    def invert(self):
        def rinvert(root):
            if root is None:
                return None
            
            # Swap
            temp = root.left
            root.left = root.right
            root.right = temp
            
            rinvert(root.left)
            rinvert(root.right)
            
            return root

        self.root = rinvert(self.root)
    
    # Validating a tree is BST or Not
    # there are 3 cases that should match
    # 1. left item ->  -inf < left node < tree node
    # 2. right item ->  tree node < right item < tree node
    # 3. left subtree -> Parent Node < left node < tree node
    
    # https://youtu.be/s6ATEkipzow?si=ZRU6haXas0qQA1y4
    def isvalid(self):
        def risvalid(root, left, right):
            if not root:
                return True
            
            if not (left < root.item < right):
                return False

            return risvalid(root.left, left, root.item) and risvalid(root.right, root.item, right)

        return risvalid(self.root, float("-inf"), float("inf"))
```

```python
tree = BST()
lst = [1,2,2,3,4,4,3]
tree.insertValues(lst)
tree.inorder()
tree.depthNodes()
tree.depthTree()
# tree.invert()
print(tree.inorder())
tree.isvalid()
```
