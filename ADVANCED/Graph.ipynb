# **Graph**

## Using Dictionary / Lists

```python
class Graph:
    def __init__(self, nov):
        self.num_vertex = nov
        # set number of key:list pairs in dictionary
        self.adj_list = {v: [] for v in range(self.num_vertex)}
    
    def add(self, u, v, weight=1):
        if 0 <= u < self.num_vertex and 0 <= v < self.num_vertex:
            self.adj_list[u].append((v, weight))
            self.adj_list[v].append((u, weight))
        else:
            print(" Invalid Vertex ")
            
    def remove(self, u, v):
        if 0 <= u < self.num_vertex and 0 <= v < self.num_vertex:
            # FIXED: We check 'vertex' against v/u, but keep the 'weight'
            self.adj_list[u] = [(vertex, weight) for vertex, weight in self.adj_list[u] if vertex != v]
            self.adj_list[v] = [(vertex, weight) for vertex, weight in self.adj_list[v] if vertex != u]
        else:
            print(" Invalid Vertex ")
            
    def has_edge(self, u, v):
        if 0 <= u < self.num_vertex and 0 <= v < self.num_vertex:
            # FIXED: Since adj_list[u] contains tuples (node, weight), 
            # we must unpack them to check if v is present.
            return any(vertex == v for vertex, weight in self.adj_list[u])
        else:
            return False
            
    def show(self):
        for key, neighbors in self.adj_list.items():
            print(f"{key} : {neighbors}")
```

```python
g = Graph(5)
g.add(0,1)
g.add(1,2)
g.add(1,3)
g.add(2,3)
g.add(3,4)
g.remove(0,1)
# g.has_edge(0,1)
g.show()
```

## Using Adjacency Matrix

```python
class Graph:
    def __init__(self, num_of_vertex=0):
        self.nov = num_of_vertex
        # Creates a 2D list (Matrix) of size nov x nov filled with zeros
        self.adj_matrix = [[0] * self.nov for i in range(self.nov)]
        
    def add(self, u, v, weight=1):
        # 0 <= u < self.nov ensures the index is within the matrix range
        if 0 <= u < self.nov and 0 <= v < self.nov:
            self.adj_matrix[u][v] = weight
            self.adj_matrix[v][u] = weight # This makes it undirected
        else:
            print("Invalid Vertices")
            
    def remove(self, u, v):
        # FIXED: Changed 0 < u to 0 <= u (Vertex 0 is valid!)
        if 0 <= u < self.nov and 0 <= v < self.nov:
            self.adj_matrix[u][v] = 0
            self.adj_matrix[v][u] = 0
        else:
            print("Invalid Vertices")
            
    def has_edge(self, u, v):
        return self.adj_matrix[u][v] != 0
    
    def show(self):
        # Improved: Prints row by row so it looks like a matrix
        for row in self.adj_matrix:
            print(" ".join(map(str, row)))
            
    def bfs(self,start_node):
        
        # check node within the range of the matrix
        if 0 <= start_node < self.nov:
            
            visited = [False]*self.nov
            queue = []
            result = []
            
            visited[start_node] = True
            queue.append(start_node)
            
            while queue:
                
                current = queue.pop(0)
                result.append(current)
                
                # searching neighbors
                for neighbour in range(self.nov):
                    if self.adj_matrix[current][neighbour] != 0 and not visited[neighbour]:
                        visited[neighbour] = True
                        queue.append(neighbour)
                        
            return result
```

```python
g = Graph(5)
g.add(0,1)
g.add(1,2)
g.add(1,3)
g.add(2,3)
g.add(3,4)
g.show()
g.bfs(0)
```
