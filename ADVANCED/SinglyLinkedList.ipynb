```python
# Creating Node
class Node:
    def __init__(self, item=None, next=None):
        self.item = item
        self.next = next

# Iterator class
class SLLIterator:
    def __init__(self, start):
        self.current = start
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if not self.current:
            raise StopIteration
        data = self.current.item
        self.current = self.current.next
        return data

# Defining a Singly Linked List
class SLL:
    def __init__(self, start=None):
        self.start = start

    # Is list Empty
    def isEmpty(self):
        return self.start == None

    # Insert item at Start
    def insertAtStart(self, item):
        self.start = Node(item, self.start)

    # Insert item at End
    def insertAtEnd(self, item):
        n = Node(item)
        if not self.isEmpty():
            temp = self.start
            while temp.next is not None:
                temp = temp.next
            temp.next = n
        else:
            self.start = n

    # Print Items
    def showList(self):
        if not self.isEmpty():
            temp = self.start
            print(" Start ", end="-->")
            while temp is not None:
                print(f" [{temp.item}] ", end="-->")
                temp = temp.next
            print(" None ")
        else:
            print("List is Empty") 

    # Searching
    def search(self, item):
        if not self.isEmpty():
            temp = self.start
            while temp is not None:
                if temp.item == item:
                    return temp
                temp = temp.next
        return None

    # Searching and returning Index
    def indexOf(self, item):
        if not self.isEmpty():
            count = 0
            temp = self.start
            while temp is not None:
                if temp.item == item:
                    return count # Return integer index
                count += 1
                temp = temp.next
        return -1 # Return -1 if not found

    # Inserting an Element after an Item
    def insertAfter(self, after, item):
        temp = self.search(after)
        if temp is not None:
            temp.next = Node(item, temp.next)

    # Deleting First Element
    def deleteFirst(self):
        if self.start is not None:
            self.start = self.start.next

    # Deleting Last Element
    def deleteLast(self):
        if self.start is None:
            pass
        elif self.start.next is None:
            self.start = None
        else:
            temp = self.start
            while temp.next.next is not None:
                temp = temp.next
            temp.next = None

    # Deleting a particular item
    def deleteItem(self, data):
        if self.start is None:
            return
        
        if self.start.item == data:
            self.start = self.start.next

        dummy = Node(0)
        dummy.next = self.start
        current = dummy
        
        while current.next is not None:
            if current.next.item == data:
                # delete and create link
                current.next = current.next.next
            else:
                # move to next node for cases like [1,2,3,3,5]
                current = current.next
            
        head = dummy.next
        
    # Iterable method
    def __iter__(self):
        return SLLIterator(self.start)
    
    # Reversing a singly linked list
    def reverse(self):
        if self.start is not None:
            current = self.start
            prev = None
            while current is not None:
                next_node = current.next   # Renamed to avoid conflict with built-in next()
                current.next = prev        
                prev = current             
                current = next_node        
            self.start = prev

    # --- Sorting Logic ---

    # Main sort method called by user
    def sort(self):
        self.start = self.merge_sort(self.start)
        
    # Helper: Recursive Merge Sort
    def merge_sort(self, head):
        # Base case: 0 or 1 node
        if not head or not head.next:
            return head
            
        mid = self.findMid(head) 
        left_half = head    
        right_half = mid.next     
        mid.next = None # Break the link
            
        left = self.merge_sort(left_half)
        right = self.merge_sort(right_half)
        
        return self.merge(left, right)
    
    # Helper: Find Middle Node
    def findMid(self, head):
        slow = head
        fast = head.next
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        return slow

    # Helper: Merge two sorted lists
    def merge(self, left, right):
        dummy = Node(0)
        current = dummy
        
        while left and right:
            if left.item < right.item:
                current.next = left
                left = left.next
            else:
                current.next = right
                right = right.next
            current = current.next
            
        current.next = left or right
        return dummy.next
    
    # Add + and create Popped elements of two lists
    def add(self,l1,l2):
        
        # create a new list
        dummy = Node(0)
        current = dummy
        carry = 0
        
        while l1 is not None or l2 is not None or carry != 0:
            
            # fatching values from head
            val1 = l1.item if l1 is not None else 0
            val2 = l2.item if l2 is not None else 0
            
            # adding values
            total = val1 + val2 + carry
            
            # holding the reminder last digit of the sum
            carry = total // 10
            
            # adding the last digit of the to the new LL
            current.next = Node(total%10)
            
            # moving to next position
            current = current.next
            
            if l1 is not None:
                l1 = l1.next
                
            if l2 is not None:
                l2 = l2.next
                
        return dummy.next
```
