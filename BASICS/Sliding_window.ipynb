### Longest Substring with At Most K Distinct Characters

```python
def max_distinct(string1):
    l = 0
    max_lenth = 0
    sset = set()
    n = len(string1)
    
    for r in range(n):
        while string1[r] in sset:
            sset.remove(string1[l])
            l += 1
            
        w = ( r - l ) + 1
        max_lenth = max(max_lenth,w)
        sset.add(string1[r])
        
    return max_lenth
        
string1 = "addaaadsfesdsd"
max_distinct(string1)
```

### Maximum average of sub array

```python
def avg_subarray(array,k):
    current_sum = 0
    max_avg = float("-inf")
    n = len(array)
    for i in range(k):
        current_sum += array[i]
        
    max_avg = current_sum / k
    
    for j in range(k,n):
        current_sum += array[j]
        current_sum -= array[j-k]
        
        avg = current_sum / k
        max_avg = max(max_avg,avg)
    
    return max_avg
array = [1,2,3,4,5,6,7,8,2,1,-1,100,-200,0,1,-100,-1000,-2000]
k = 4
avg_subarray(array,k)
```
