### Return array of squred numbers into non decresing order (may have duplicate values)

```python
def squared_array(nums):
    lst = []
    l = 0
    r = len(nums) - 1
    
    while l <= r:
        if abs(nums[l]) > abs(nums[r]):
            lst.append(nums[l]**2)
            l += 1
        else:
            lst.append(nums[r]**2)
            r -= 1
    
    lst.reverse()
            
    return lst

nums = [-4,-1,0,3,10]
squared_array(nums)
```

### Reverse string in place

```python
def reverse_lst(s):
    l = 0
    r = len(s) - 1
    
    while l<=r:
        s[l],s[r] = s[r],s[l]
        l += 1
        r -= 1
        
    return s
s = ["h","e","l","l","o"]
reverse_lst(s)
```

### 2 sum with constant space complexity

```python
def two_sum_constant(num,target):
    sum = 0
    l = 0
    r = len(num) - 1
    
    while l<=r:
        sum = num[l]+num[r]
        if sum == target:
            return [l+1,r+1]
        elif sum < target:
            l+=1
        else:
            r-=1       
    
    
# num = [2,7,11,15]
# target = 9

num = [2,3,4]
target = 6
two_sum_constant(num,target)
```

### Check a string is PALLINDROME or not ( Ignore no alphanumeric characters)

```python
def pallindrome_alnum(str):
    l = 0
    r = len(str) - 1
    
    while l<r:
        
        if not str[l].isalnum():
            l += 1
            continue
        
        if not str[r].isalnum():
            r += 1
            continue
        
        if str[l].lower() != str[r].upper():
            return False
        
        l += 1
        r -= 1
        
    return True

str = "race a car"
pallindrome_alnum(str)
```
