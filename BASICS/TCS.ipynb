1. Find the maximum product possible by multiplying any subset of elements (can include negative, zero , positive )

```python
# loop through each element , ignore 0 , find the max negative element , divide the Product 
# by max negative element if total number of negative elements is Odd

def max_product(array):
    count_even_neg = 0
    max_negative = float("-inf")
    product = 1
    
    for i in array:
        if i!= 0:
            if i < 0:
                count_even_neg += 1
                max_negative = max(max_negative,i)
            product *= i

    if count_even_neg % 2 != 0:
        return product / max_negative
    
    return product

array = [-5, -4, -3, -2, 0, 0]
max_product(array)
```

2. You need to calculate the parking fee based on the number of hours parked.
    Rules:
    1. First 2 hours -> $100 per hour
    2. Next 3 hours (3rd to 5th hour) -> $50 per hour
    3. Beyond 5 hours -> $20 per hour

```python
def park_fee(num):
    fee = 0
    try:
        num = int(num)
        if num <= 0:
            return 0
        elif num < 2:
            return 2 * 100
        elif num < 5:
            return 2*50 + 3*100
        else:
            return 2*100 + 3*50 + (num-5)*20
    except ValueError:
        return "Error"
    
num = input(" Enter a number ")
park_fee(num)
```
