## Converting digits to an Integer and visa versa

```python
# converting list to a number

lst = [1,2,3,4,5]

digit = 0
for i in lst:
    digit = digit * 10 + i

print(digit)
```

```python
# converting digit to list / reversing

lst = []
digit = 0
total = 12345

while total>0:
    
    # get the reminder
    reminder = total % 10
    
    lst.append(reminder)
    # hold the reminder and merge with nxt reminder
    
    digit = digit * 10 + reminder
    
    # remove the last digit
    
    total = total // 10
    
print(lst)
print(digit)
```

```python
# GCD / HCF

def GCD(a,b):
    if b == 0:
        return a
    return GCD(b,a%b)

GCD(100,10)
```
