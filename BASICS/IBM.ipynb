```python
def fix_time(time_str):
    # ---------------------------------------------------------
    # 1. FIND THE MINIMUM TIME
    # Min time is simple: every hidden digit '@' becomes '0'
    # ---------------------------------------------------------
    min_time = time_str.replace('@', '0')
    
    # ---------------------------------------------------------
    # 2. FIND THE MAXIMUM TIME
    # We must convert the string to a list so we can modify it
    # ---------------------------------------------------------
    max_t = list(time_str)
    
    # Fix Hours (Index 0 and 1) -> Format HH:MM:SS
    if max_t[0] == '@':
        # If the second digit is 4-9, the first digit can only be 1 (e.g., 19 is valid, 29 is not)
        if max_t[1] != '@' and int(max_t[1]) > 3:
            max_t[0] = '1'
        else:
            max_t[0] = '2'
            
    if max_t[1] == '@':
        # If the first digit is 2, the highest second digit is 3 (for 23:xx:xx)
        if max_t[0] == '2':
            max_t[1] = '3'
        else:
            max_t[1] = '9'
            
    # Fix Minutes (Index 3 and 4)
    if max_t[3] == '@':
        max_t[3] = '5' # Max tens place for minutes is 5 (59)
    if max_t[4] == '@':
        max_t[4] = '9'
        
    # Fix Seconds (Index 6 and 7)
    if max_t[6] == '@':
        max_t[6] = '5' # Max tens place for seconds is 5 (59)
    if max_t[7] == '@':
        max_t[7] = '9'
        
    # Join the list back into a string
    max_time = "".join(max_t)
    
    # ---------------------------------------------------------
    # 3. PRINT RESULTS
    # ---------------------------------------------------------
    print(f"Original String : {time_str}")
    print(f"Minimum Time    : {min_time}")
    print(f"Maximum Time    : {max_time}")
    print("-" * 30)

# --- Test Cases ---
fix_time("10:00:5@")
fix_time("@9:@@:@@")
fix_time("2@:1@:00")
fix_time("@@:@@:@@")
```

```python
def license(string_id):
    # Alphanumneri
    if string_id.isalnum() and 2 <=len(string_id) <= 10 and any(i.isdigit() for i in string_id):
        new_string = string_id.upper()
        # check divisble by 3
        num = len(new_string)//3
        result = ""
        if num >= 0:
            # grouping of max 3
            count = 0
            for i in string_id:
                if count == 3:
                    result += "-"
                    count = 0
                result += i
                count += 1
            return result
    else:
        print("INVALID")

id = "ABC123XYZ"
license(id)
```

```python
def pro_cost(lst,k):
    unique_costs = set(lst)
    count = 0    
    for cost in unique_costs:
        if ( cost + k ) in unique_costs:
            count += 1
            
    return count

lst = [4, 2, 1, 6, 2, 4, 5, 5, 3]
k = 2
pro_cost(lst,k)
```

```python
strings = "a3b2c5a1g9c2"
n = len(strings)
str_map = {}
for i in range(0,n,2):
    if strings[i] not in str_map:
        str_map.update({strings[i]:int(strings[i+1])})
    else:
        str_map[strings[i]] += int(strings[i+1])

# sort based on values
sorted_map = dict(sorted(str_map.items() ,key= lambda x: x[1]))
        
result = "".join(f"{char}{num}" for char,num in sorted_map.items())

print(result)
```

```python
# Paranthesis Flags

# exp 1 : ["())","())(()","(","(((("," "]
# Output : [0,1,0,0]

def splitter(str):
    if str is None:
        return 1
    return 1 if str.count("(") == str.count(")") else 0

def output(lst):
    return [splitter(i) for i in lst]

a = ["())","())(()","(","((((",""]
output(a)
```

```python
# missing numbers
# exp 1 : 1 3 2 5 5 
# output : 4

# exp 1 : 2 3 5 2 4
# output : 1

def miss_num(lst):
    n = len(lst)
    flag = [0]*n
    
    for num in lst:
        if 1 <= num <= n:
            flag[num - 1] = 1
    
    for i in range(n):
        if flag[i] == 0:
            return i+1
    return -1
        
print(miss_num([1, 3, 2, 5, 5])) # Output: 4
print(miss_num([2, 3, 5, 2, 4])) # Output: 1
```
